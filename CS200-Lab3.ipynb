## Exercise 1 — The `while` Loop (zyBook 4.1)

This is the investment program from zyBook 4.1. It counts how many years it takes for an investment to double at 5% interest. Run it.

````python
RATE = 5.0
INITIAL_BALANCE = 10000.0
TARGET = 2 * INITIAL_BALANCE

balance = INITIAL_BALANCE
year = 0

while balance < TARGET :
   year = year + 1
   interest = balance * RATE / 100
   balance = balance + interest

print("The investment doubled after", year, "years.")
````
